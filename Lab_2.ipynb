```python
def create_tic_tac_toe_matrix(n):

    matrix = []
    for _ in range(n):
        matrix.append([0] * n)
    return matrix
```

```python
def draw_game_board(board_matrix):

    n = len(board_matrix)
    m = len(board_matrix[0]) if n > 0 else 0


    col_labels = '   '.join([chr(65 + j) for j in range(m)])
    print(f'    {col_labels}')

    horizontal_line = ' ---' * m + ' '

    for i in range(n):
        print(f'  {horizontal_line}')
        row_str = f'{i + 1} |'
        for j in range(m):
            cell_value = board_matrix[i][j]
            if cell_value == 0:
                row_str += '   |'
            elif cell_value == 1:
                row_str += ' X |'
            elif cell_value == 2:
                row_str += ' O |'
            else:
                row_str += ' ? |'
        print(row_str)
    print(f'  {horizontal_line}')
```

```python
def check_game_status(board_matrix):

    n = len(board_matrix)

    for row in board_matrix:
        if all(cell == row[0] and cell != 0 for cell in row):
            return row[0]

    for col_idx in range(n):
        first_cell_in_col = board_matrix[0][col_idx]
        if first_cell_in_col != 0 and all(board_matrix[row_idx][col_idx] == first_cell_in_col for row_idx in range(n)):
            return first_cell_in_col


    first_diag_cell = board_matrix[0][0]
    if first_diag_cell != 0 and all(board_matrix[i][i] == first_diag_cell for i in range(n)):
        return first_diag_cell


    first_anti_diag_cell = board_matrix[0][n-1]
    if first_anti_diag_cell != 0 and all(board_matrix[i][n-1-i] == first_anti_diag_cell for i in range(n)):
        return first_anti_diag_cell

    for row in board_matrix:
        if 0 in row:
            return -1
    return 0
```

problem 5

```python
def place_player_move(board_matrix, player_number, row, col):

    n = len(board_matrix)


    if not (0 <= row < n and 0 <= col < n):
        print(f"Invalid move: ({row}, {col}) is out of bounds for a {n}x{n} board.")
        return False

    if board_matrix[row][col] == 0:
        board_matrix[row][col] = player_number
        return True
    else:
        print(f"Invalid move: Position ({row}, {col}) is already taken by player {board_matrix[row][col]}.")
        return False
```

```python
def make_move_from_string(board_matrix, player_number, location_string):

    if not isinstance(location_string, str) or len(location_string) < 2:
        print("Invalid location string format. Expected format: 'A1' or 'C3'.")
        return False

    col_char = location_string[0].upper()
    row_str = location_string[1:]

    n = len(board_matrix)

    try:
        col = ord(col_char) - ord('A')
        row = int(row_str) - 1
    except ValueError:
        print("Invalid location string format. Row must be a number.")
        return False

    if not (0 <= col < n):
        print(f"Invalid column '{col_char}'. Column must be between A and {chr(ord('A') + n - 1)}.")
        return False

    return place_player_move(board_matrix, player_number, row, col)
```

```python
def get_and_make_player_move(board_matrix, player_number):
  layer_number (int): The current player (1 for 'X', 2 for 'O').

    player_char = 'X' if player_number == 1 else 'O'
    while True:
        location_string = input(f"Player {player_number} ({player_char}), enter your move (e.g., A1): ")
        if make_move_from_string(board_matrix, player_number, location_string):
            print(f"Player {player_number} ({player_char}) successfully placed at {location_string.upper()}.")
            break
        else:
            print("Please try again.")
```

```python
def play_tic_tac_toe_game(n=3):

    board = create_tic_tac_toe_matrix(n)
    current_player = 1

    print("\n--- Starting Tic-Tac-Toe Game ---")

    while True:
        draw_game_board(board)

        get_and_make_player_move(board, current_player)

        game_status = check_game_status(board)

        if game_status == 1:
            draw_game_board(board)
            print("\n!!! Player 1 (X) wins! Congratulations !!!")
            break
        elif game_status == 2:
            draw_game_board(board)
            print("\n!!! Player 2 (O) wins! Congratulations !!!")
            break
        elif game_status == 0:
            draw_game_board(board)
            print("\n--- It's a draw! Good game! ---")
            break

        current_player = 2 if current_player == 1 else 1
```

problem 11

```python
import random

def get_computer_move(board_matrix, computer_player_number):
    n = len(board_matrix)
    opponent_player_number = 1 if computer_player_number == 2 else 2

    empty_cells = []
    for r in range(n):
        for c in range(n):
            if board_matrix[r][c] == 0:
                empty_cells.append((r, c))


    for r, c in empty_cells:
        board_matrix[r][c] = computer_player_number
        if check_game_status(board_matrix) == computer_player_number:
            board_matrix[r][c] = 0
            print(f"Computer (Player {computer_player_number}) chooses {chr(ord('A') + c)}{r + 1} (winning move).")
            return (r, c)
        board_matrix[r][c] = 0


    for r, c in empty_cells:
        board_matrix[r][c] = opponent_player_number
        if check_game_status(board_matrix) == opponent_player_number:
            board_matrix[r][c] = 0
            print(f"Computer (Player {computer_player_number}) chooses {chr(ord('A') + c)}{r + 1} (blocking move).")
            return (r, c)
        board_matrix[r][c] = 0


    if n % 2 == 1:
        center_coord = n // 2
        if board_matrix[center_coord][center_coord] == 0:
            print(f"Computer (Player {computer_player_number}) chooses {chr(ord('A') + center_coord)}{center_coord + 1} (center).")
            return (center_coord, center_coord)


    corners = [(0, 0), (0, n - 1), (n - 1, 0), (n - 1, n - 1)]
    random.shuffle(corners)
    for r, c in corners:
        if board_matrix[r][c] == 0:
            print(f"Computer (Player {computer_player_number}) chooses {chr(ord('A') + c)}{r + 1} (corner).")
            return (r, c)


    if empty_cells:
        r, c = random.choice(empty_cells)
        print(f"Computer (Player {computer_player_number}) chooses {chr(ord('A') + c)}{r + 1} (random).")
        return (r, c)

    return None
```

```python
def play_tic_tac_toe_computer_game(n=3):
    board = create_tic_tac_toe_matrix(n)


    while True:
        player_choice = input("Do you want to be Player 1 (X) or Player 2 (O)? Enter 1 or 2: ")
        if player_choice == '1':
            human_player = 1
            computer_player = 2
            break
        elif player_choice == '2':
            human_player = 2
            computer_player = 1
            break
        else:
            print("Invalid choice. Please enter 1 or 2.")

    current_player = 1

    print(f"\n--- Starting Tic-Tac-Toe Game (Human: Player {human_player}, Computer: Player {computer_player}) ---")

    while True:
        draw_game_board(board)

        if current_player == human_player:
            print(f"\nHuman Player {human_player}'s turn:")
            get_and_make_player_move(board, human_player)
        else:
            print(f"\nComputer Player {computer_player}'s turn:")
            row, col = get_computer_move(board, computer_player)
            place_player_move(board, computer_player, row, col)

        game_status = check_game_status(board)

        if game_status == human_player:
            draw_game_board(board)
            print(f"\n!!! Human Player {human_player} ({'X' if human_player == 1 else 'O'}) wins! Congratulations !!!")
            break
        elif game_status == computer_player:
            draw_game_board(board)
            print(f"\n!!! Computer Player {computer_player} ({'X' if computer_player == 1 else 'O'}) wins! Better luck next time !!!")
            break
        elif game_status == 0:
            draw_game_board(board)
            print("\n--- It's a draw! Good game! ---")
            break

        current_player = 2 if current_player == 1 else 1
```

```python
def get_random_computer_move(board_matrix, computer_player_number):
    n = len(board_matrix)
    empty_cells = []
    for r in range(n):
        for c in range(n):
            if board_matrix[r][c] == 0:
                empty_cells.append((r, c))

    if empty_cells:
        r, c = random.choice(empty_cells)
        print(f"Computer (Player {computer_player_number}) chooses {chr(ord('A') + c)}{r + 1} (random).")
        return (r, c)
    return None
```

```python
def play_tic_tac_toe_computer_vs_computer_game(n=3, player1_strategy='smart', player2_strategy='random', print_game=True):
    board = create_tic_tac_toe_matrix(n)
    current_player = 1

    if print_game:
        print(f"\n--- Starting Computer vs. Computer Game (P1: {player1_strategy}, P2: {player2_strategy}) --- for {n}x{n} board")

    while True:
        if print_game:
            draw_game_board(board)

        move = None
        if current_player == 1:
            if player1_strategy == 'smart':
                move = get_computer_move(board, 1)
            else:
                move = get_random_computer_move(board, 1)
        else:
            if player2_strategy == 'smart':
                move = get_computer_move(board, 2)
            else:
                move = get_random_computer_move(board, 2)

        if move:
            place_player_move(board, current_player, move[0], move[1])
        else:

            break

        game_status = check_game_status(board)

        if game_status == 1:
            if print_game:
                draw_game_board(board)
                print("!!! Player 1 wins! !!!")
            return 1
        elif game_status == 2:
            if print_game:
                draw_game_board(board)
                print("!!! Player 2 wins! !!!")
            return 2
        elif game_status == 0:
            if print_game:
                draw_game_board(board)
                print("--- It's a draw! ---")
            return 0

        current_player = 2 if current_player == 1 else 1
```

machine vs machine test

```python
num_games = 10
board_sizes = [3, 4, 5]

results = {}

for n in board_sizes:
    print(f"\n--- Simulating {num_games} games on a {n}x{n} board (Smart P1 vs. Random P2) ---")
    smart_wins = 0
    random_wins = 0
    draws = 0

    for i in range(num_games):
        print(f"Game {i+1}:")
        winner = play_tic_tac_toe_computer_vs_computer_game(n, player1_strategy='smart', player2_strategy='random', print_game=False)
        if winner == 1:
            smart_wins += 1
        elif winner == 2:
            random_wins += 1
        else:
            draws += 1

    results[n] = {
        'smart_wins': smart_wins,
        'random_wins': random_wins,
        'draws': draws,
        'smart_win_rate': (smart_wins / num_games) * 100
    }
    print(f"Results for {n}x{n} board:")
    print(f"  Smart Player 1 Wins: {smart_wins}")
    print(f"  Random Player 2 Wins: {random_wins}")
    print(f"  Draws: {draws}")
    print(f"  Smart Player 1 Win Rate: {results[n]['smart_win_rate']:.2f}%")

print("\n--- Final Simulation Results ---")
for n, res in results.items():
    print(f"Board Size {n}x{n}: Smart Win Rate = {res['smart_win_rate']:.2f}%")
```
